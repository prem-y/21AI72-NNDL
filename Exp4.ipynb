```python
import tensorflow as tf
from tensorflow.keras import layers, models, regularizers
from tensorflow.keras.datasets import fashion_mnist
```

```python
(x_train, y_train), (x_test, y_test) = fashion_mnist.load_data()
x_train, x_test = x_train/255.0 , x_test/255.0

y_train = tf.keras.utils.to_categorical(y_train, 10)
y_test = tf.keras.utils.to_categorical(y_test, 10)

# Base Model
model1 = models.Sequential([
    layers.Input(shape=(28,28,1)),
    layers.Conv2D(32, (3,3), activation='relu'),
    layers.MaxPooling2D(),
    layers.Conv2D(64, (3,3), activation='relu'),
    layers.MaxPooling2D(),
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dense(10, activation='softmax')
])
model1.compile(optimizer='adam',
               loss='categorical_crossentropy',
               metrics=['accuracy'])
model1.fit(x_train, y_train, epochs=5, batch_size=32, verbose=2)
test_loss, test_acc = model1.evaluate(x_test, y_test, verbose=2)

print(f"Test Accuracy: {test_acc:.4f}")
```

```python
# model with L1 regularization
model2 = models.Sequential([
    layers.Input(shape=(28,28,1)),
    layers.Conv2D(32, (3,3), activation='relu'),
    layers.MaxPooling2D(),
    layers.Conv2D(64, (3,3), activation='relu'),
    layers.MaxPooling2D(),
    layers.Flatten(),
    layers.Dense(128, activation='relu', kernel_regularizer=regularizers.l1(1e-4)),
    layers.Dense(10, activation='softmax')
])

model2.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
model2.fit(x_train,y_train, epochs=5, batch_size=32, verbose=2)
```

```python
train_loss,train_acc = model2.evaluate(x_train, y_train, verbose=2)
test_loss,test_acc = model2.evaluate(x_test, y_test, verbose=2)

print(f"Train accuracy : {train_acc:.2f}")
print(f"Test accuracy: {test_acc:.2f}")
```

```python
# model with L2 regularization
model3 = models.Sequential([
    layers.Input(shape=(28,28,1)),
    layers.Conv2D(32, (3,3), activation='relu'),
    layers.MaxPooling2D(),
    layers.Conv2D(64, (3,3), activation='relu'),
    layers.MaxPooling2D(),
    layers.Flatten(),
    layers.Dense(128, activation='relu', kernel_regularizer=regularizers.l2(1e-4)),
    layers.Dense(10, activation='softmax')
])
model3.compile(optimizer='adam', loss = 'categorical_crossentropy', metrics=['accuracy'])
model3.fit(x_train, y_train, batch_size=32, epochs=5, verbose=2)
```

```python
train_loss,train_acc = model3.evaluate(x_train, y_train, verbose=2)
test_loss,test_acc = model3.evaluate(x_test, y_test, verbose=2)

print(f"Train accuracy : {train_acc:.2f}")
print(f"Test accuracy: {test_acc:.2f}")
```

```python
# Model with dropout
model4 = models.Sequential([
    layers.Input(shape=(28,28,1)),
    layers.Conv2D(32, (3,3), activation='relu'),
    layers.MaxPooling2D(),
    layers.Conv2D(64, (3,3), activation='relu'),
    layers.MaxPooling2D(),
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dropout(0.5),
    layers.Dense(10, activation='softmax')
])

model4.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
model4.fit(x_train, y_train, batch_size=32, epochs=5, verbose=2)
```

```python
train_loss,train_acc = model4.evaluate(x_train, y_train, verbose=2)
test_loss,test_acc = model4.evaluate(x_test, y_test, verbose=2)

print(f"Train accuracy : {train_acc:.2f}")
print(f"Test accuracy: {test_acc:.2f}")
```
